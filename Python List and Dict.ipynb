```python
#sets
my_set={'a','e','i','o','u','a','a','i'}
print(my_set)

```

#Explanation for setoutput
- A set does not allow duplicate values.Therefore,the duplicate values 'a' and 'i' are removed automatically.
- The output contains only unique values.
- Sets are unordered so order of elements may vary in the output.

  

```python
my_set[4]='s'
```

#Explanation of TypeError
- Sets do not support indexing or item assignment because they are unordered.
- Therefore, my_set[4]= 's' gives a 'TypeError'.
- To add a new element to a set, we can use the add() method.

```python
set1 = {1,3,5,7,9}
set2 = {2,3,5,8,10}
```

```python
set_union=set1.union(set2)
print("Union:",set_union)
```

```python
set_intersection=set1.intersection(set2)
print("Intersection:",set_intersection)
```
