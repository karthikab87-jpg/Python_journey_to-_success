```python
# 1.Create 1D Numpy Array

import numpy as np

temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_w1
```

```python
#2.Inspection and Properties

print('Shape:',temperatures_w1.shape)
print('Data Type:',temperatures_w1.dtype)
print('Number of Elements:',temperatures_w1.size)

temperatures_w1.shape
```

```python
# 3.Array operations

fahrenheit = (temperatures_w1 * 9/5) + 32
print(fahrenheit)

```

```python
# find max,min and mean temperatures

print(np.max(temperatures_w1))
print(np.min(temperatures_w1))
print(round(np.mean(temperatures_w1),2))
```

```python
#4. Array slicing and indexing

temperatures_w1[:3]
```

```python
temperatures_w1[-2:]
```

```python
temperatures_w1[2:5]
```

```python
# 5.2D array

temperatures_inweeks=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
                             [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(temperatures_inweeks)
```

```python
# Inspect and slice the 2D array

print('Shape:',temperatures_inweeks.shape)
print('Data Type:',temperatures_inweeks.dtype)
print('Total no:of elements:',temperatures_inweeks.size)
```

```python
# Extract each week 

print(temperatures_inweeks[0])
print(temperatures_inweeks[1])


```

```python
# Extract temperatures for weeks and weekends 

print(temperatures_inweeks[:,-2:])



```

Pandas DataFrame


```python
# 1.Creating Pandas Data Frame

transactions=pd.DataFrame({'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
'ProductCategory':['Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West',
'South', 'North'],
'Amount':[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]})

transactions
```

```python
#2. Data Exploration

print(transactions)
```

```python
print(transactions.info())
```

```python
# 2.2

print(transactions[['ProductCategory', 'Amount']])

```

```python
# 2.3

print(transactions.iloc[:, -3:])
```

```python
# 2.4

print(transactions[(transactions['Region']=='North') & (transactions['Amount']>200)])
```

```python
# 2.5

print(transactions['ProductCategory'].value_counts())
```

```python
# 2.6

print(transactions['Region'].unique())
```

```python
# 2.7

print(transactions.groupby('Region')['Amount'].mean())
```

```python
# Manipulating the dataframe

transactions.loc[transactions['TransactionID']==102,'Amount']= 165
print(transactions)
```

```python
# Add a new column

transactions['Discount'] = transactions['Amount'] * 0.10

transactions

```

```python
# Remove 

transactions=transactions[transactions['TransactionID']!=109]
transactions
```

```python
# Deletion

transactions=transactions.drop('Discount',axis=1)
```
